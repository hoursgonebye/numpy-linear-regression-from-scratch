```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
strength = np.array([1, 2, 3, 4])
actual_damage = np.array([5, 8, 11, 14])

weight = 1
bias = 0



predictions = weight * strength + bias
errors = predictions - actual_damage

mse = (errors ** 2).mean()

print("Predictions:", predictions)
print("Errors:", errors)
print("Mean squared error:", mse)
```

```python
strength = np.array([1, 2, 3, 4])
actual_damage = np.array([6, 7, 12, 13])

weight = 2.0
bias = 1.0
learning_rate = 0.01

for i in range(1000):
    predictions = strength * weight + bias
    errors = predictions - actual_damage 

    weight_gradient = 2 * (errors * strength).mean()
    bias_gradient = 2 * errors.mean()

    weight -= learning_rate * weight_gradient
    bias -= learning_rate * bias_gradient

    



print("Final Weight:", weight, "Final Bias:", bias)
```

```python
new_strength = 5 #test model
predicted_damage = weight * new_strength + bias

print("Predicted damage:", predicted_damage)
```

```python
test_strength = np.array([5, 6, 7]) #test model on non trained data
test_damage = np.array([17, 20, 23])

test_predictions = weight * test_strength + bias
test_mse = ((test_predictions - test_damage) ** 2).mean()

print("Test predictions:", test_predictions)
print("Test MSE:", test_mse)
```

```python
final_predictions = weight * strength + bias
final_mse = ((final_predictions - actual_damage) ** 2).mean()

print("Predictions:", final_predictions)
print("Final MSE:", final_mse)
```

```python
plt.scatter(strength, actual_damage, label="Observed Damage")
plt.plot(strength, final_predictions, color="orange", label="Model predictions")

plt.xlabel("Strength")
plt.ylabel("Damage")
plt.title("A straight-line model fitting noisy data")
plt.legend()
plt.show()
```

```python
baseline_prediction = actual_damage.mean()
baseline_mse = ((baseline_prediction - actual_damage) ** 2).mean()
baseline_test_mse = ((baseline_prediction - test_damage) ** 2).mean()

print("Baseline MSE:", baseline_mse)
print("Baseline test MSE:", baseline_test_mse)
print("Model MSE:", final_mse)
print("Model test MSE:", test_mse)
```
